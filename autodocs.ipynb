Execução auditável de `PRO-CEPRAEA-DOC-CONS-001 v1.1.0`. A implementação canônica está em `scripts/u01_inventory.py` e usa somente a biblioteca padrão.

```python
import os
import socket
import sys
from pathlib import Path

source = Path("/home/davis/cepraea-docs/docs/reconciliação")
output = Path("/home/davis/cepraea-docs/artifacts/PRO-CEPRAEA-DOC-CONS-001/U01")

print("Python:", sys.executable)
print("Servidor:", socket.gethostname())
print("Colab:", os.environ.get("COLAB_RELEASE_TAG", "Não identificado"))
print("Pasta acessível:", source.is_dir() and os.access(source, os.R_OK | os.X_OK))
print("Origem:", source)
print("Saída separada:", output)
```

```python
import subprocess

inventory = subprocess.run(
    [
        sys.executable,
        "scripts/u01_inventory.py",
        "--source", str(source),
        "--output", str(output),
        "--generated-at", "2026-10-16T22:15:33-03:00",
    ],
    check=True, text=True, capture_output=True,
)
print(inventory.stdout)
```

```python
tests = subprocess.run(
    [sys.executable, "-m", "unittest", "-v", "tests.test_u01_inventory"],
    check=True, text=True, capture_output=True,
)
print(tests.stdout + tests.stderr)
```
